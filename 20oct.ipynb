```python

class Furniture:
    def __init__(self, f_id, name, price, color, quantity):
        self.f_id = f_id
        self.name = name
        self.price = price
        self.color = color
        self.quantity = quantity

    def display(self):
        print("Furniture ID:", self.f_id)
        print("Name:", self.name)
        print("Price:", self.price)
        print("Color:", self.color)
        print("Quantity:", self.quantity)


f1 = Furniture(1, "Chair", 1500, "Black", 10)
f2 = Furniture(2, "Table", 5000, "Brown", 5)
f3 = Furniture(3, "Sofa", 25000, "Grey", 3)
f4 = Furniture(4, "Bed", 18000, "White", 4)
f5 = Furniture(5, "Cupboard", 12000, "Blue", 2)

furniture_list = []

furniture_list.append(f1)
furniture_list.append(f2)
furniture_list.append(f3)
furniture_list.append(f4)
furniture_list.append(f5)

print("Displaying all furniture objects:")
for obj in furniture_list:
    obj.display()
    print()

print("\nDisplaying furniture names:")
for obj in furniture_list:
    print("Name:", obj.name)

print("\nDisplaying furniture id:")
for obj in furniture_list:
    print("Furniture ID:", obj.f_id)
    print("Furniture name:", obj.name)
    

print("\nDisplaying furniture prices:")
for obj in furniture_list:
    print("Price:", obj.price)
    print("Furniture name:", obj.name)

print("\nDisplaying furniture colors:")
for obj in furniture_list:
    print("Color:", obj.color)
    print("Furniture name:", obj.name)

print("\nDisplaying furniture quantities:")
for obj in furniture_list:
    print("Quantity:", obj.quantity)
    print("Furniture name:", obj.name)
```
